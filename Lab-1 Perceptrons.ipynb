### Lab Exercise-1
### Design perceptrons for AND, OR, NAND, NOR gates.

```python
def step_function(x):
    return 1 if x >= 0 else 0

def perceptron(x1, x2, w1, w2, b):
    net = w1*x1 + w2*x2 + b
    return step_function(net)

# Define logic gates
def AND(x1, x2):
    return perceptron(x1, x2, 1, 1, -1.5)

def OR(x1, x2):
    return perceptron(x1, x2, 1, 1, -0.5)

def NAND(x1, x2):
    return perceptron(x1, x2, -1, -1, 1.5)

def NOR(x1, x2):
    return perceptron(x1, x2, -1, -1, 0.5)

# Test all gates
inputs = [(0,0), (0,1), (1,0), (1,1)]

print("AND Gate")
for x1, x2 in inputs:
    print(x1, x2, "->", AND(x1, x2))

print("\nOR Gate")
for x1, x2 in inputs:
    print(x1, x2, "->", OR(x1, x2))

print("\nNAND Gate")
for x1, x2 in inputs:
    print(x1, x2, "->", NAND(x1, x2))

print("\nNOR Gate")
for x1, x2 in inputs:
    print(x1, x2, "->", NOR(x1, x2))
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Input points
X = np.array([
    [0, 0],
    [0, 1],
    [1, 0],
    [1, 1]
])

# Gates: name -> (w1, w2, b)
gates = {
    "AND": (1, 1, -1.5),
    "OR": (1, 1, -0.5),
    "NAND": (-1, -1, 1.5),
    "NOR": (-1, -1, 0.5)
}

def step(z):
    return 1 if z >= 0 else 0

def plot_gate(w1, w2, b, title):
    x_vals = np.linspace(-0.5, 1.5, 100)
    y_vals = (-w1 * x_vals - b) / w2  # Decision boundary

    plt.figure()

    # Plot input points
    for x in X:
        z = w1*x[0] + w2*x[1] + b
        y = step(z)
        plt.scatter(x[0], x[1])
        plt.text(x[0]+0.05, x[1]+0.05, str(y))

    # Plot boundary
    plt.plot(x_vals, y_vals)

    plt.xlabel("x1")
    plt.ylabel("x2")
    plt.title(title + " Gate Decision Boundary")
    plt.xlim(-0.5, 1.5)
    plt.ylim(-0.5, 1.5)
    plt.grid(True)
    plt.show()

# Plot all gates
for gate, params in gates.items():
    plot_gate(*params, gate)
```
