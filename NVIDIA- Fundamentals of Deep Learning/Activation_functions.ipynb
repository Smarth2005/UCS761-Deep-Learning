```python
import tensorflow as tf
```

```python
from tensorflow import keras
import matplotlib.pyplot as plt
plt.style.use("default")
```

```python
# Generate Data Points
X = tf.linspace(-10, 10, 100)
X
```

```python
X.shape
```

```python
# Sigmoid activation function
# Activations
y = tf.keras.activations.sigmoid(X)

# Visualize graph
plt.plot(X, y, label=r"$sigmoid(x) = \frac{1}{1+e^{-x}}$")
plt.title("Sigmoid")
plt.xlabel("X")
plt.ylabel("sigmoid(X)")
plt.legend()
plt.grid(True, linestyle="--", linewidth=0.7, alpha=0.6)
plt.show()
```

```python
# tanh activation function
# Activations
y = tf.keras.activations.tanh(X)

# Visualize graph
plt.plot(X, y, label="tanh(x)")
plt.title("Hyperbolic tangent")
plt.xlabel("X")
plt.ylabel("tanh(X)")
plt.legend()
plt.grid(True, linestyle="--", linewidth=0.7, alpha=0.6)
plt.show()
```

```python
# ReLU activation function
# Activations
y = tf.keras.activations.relu(X)

# Visualize graph
plt.plot(X, y, label="max(0,x)")
plt.title("ReLU")
plt.xlabel("x")
plt.ylabel("max(0,x)")
plt.legend()
plt.grid(True, linestyle="--", linewidth=0.7, alpha=0.6)
plt.show()
```

```python
# Leaky ReLU
# Activations
y = tf.nn.leaky_relu(X)

# Visualize graph
plt.plot(X, y)
plt.title("Leaky ReLU")
plt.xlabel("X")
plt.ylabel("Leaky ReLU")
plt.legend()
plt.grid(True, linestyle="--", linewidth=0.7, alpha=0.6)
plt.show()
```

```python
# Elu (Exponential linear unit)
# Activations
y = tf.keras.activations.elu(X, alpha=0.9)

# Visualize graph
plt.plot(X, y, label=r"$f(x)=\{x,\ x>0;\ \alpha(e^x-1),\ x\leq 0\}$")
plt.title("Elu")
plt.xlabel("x")
plt.ylabel("elu(x)")
plt.legend()
plt.grid(True, linestyle="--", linewidth=0.7, alpha=0.6)
plt.show()
```

```python
# Softmax
x = tf.constant([150, 50, 10], dtype=tf.float32)
x = tf.expand_dims(x, axis=0)
print(x.shape, x)

y = tf.keras.activations.softmax(x)
print(y)
```
